```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread('Coins.png')
plt.imshow(img, cmap='gray')
```

```python
img_RGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.imshow(img_RGB)
```

```python
img_blur = cv.GaussianBlur(cv.cvtColor(img, cv.COLOR_BGR2GRAY), (5, 5), 0)
#img_blur = cv.medianBlur(cv.cvtColor(img, cv.COLOR_BGR2GRAY), 5)
#img_blur = cv.bilateralFilter(cv.cvtColor(img, cv.COLOR_BGR2GRAY), 9, 75, 75)
plt.imshow(img_blur, cmap='gray')
```

```python
ret, img_threshhold = cv.threshold(img_blur, 200, 255, cv.THRESH_BINARY)
plt.imshow(img_threshhold, cmap='gray')
```

```python
img_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
img_contours = img_threshhold.copy()
contours, hierarchy = cv.findContours(img_contours, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)
for i in range(len(contours)):
    cv.drawContours(img_contours, contours, i, (0, 0, 255), 2)
plt.imshow(img_contours)
```
